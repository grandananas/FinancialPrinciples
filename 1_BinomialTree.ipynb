## **5.1. PRICING AND HEDGING VANILLA OPTIONS WITH THE BINOMIAL TREE**

In this lecture, we will learn how to write a Python function to construct a simple binomial tree.

#### Generating the price paths

```python
def binomial_tree(S_ini, T, u, d, N):
    """
    Args:
    - S_ini: Initial underlying price (S_0)
    - T: Time to maturity (time horizone in years)
    - u: upward price multiplier
    - d: downward price multiplier
    - N: Number of time steps.
    """
    S = np.zeros([N + 1, N + 1])  # Underlying price paths
    for i in range(0, N + 1):
        S[N, i] = S_ini * (u ** (i)) * (d ** (N - i)) #terminal nodes
    for j in range(N - 1, -1, -1):
        for i in range(0, j + 1):
            S[j, i] = S_ini * (u ** (i)) * (d ** (j - i)) #middle nodes
            # S[j, i] represents the price at step j (period j = 1, ... N) after i up moves and j - i down moves.
    return S
```

```python
S_paths = binomial_tree(100, 1, 1.15, 0.85, 2)
S_paths
```

### **5.1.1. European Call Option**

Strike: $K = \$ 90$;

Maturity: $T = 1$ year;

The payoff is computed at maturity (at terminal nodes).

#### Call payoff function:

```python
def binomial_tree_call(S_ini, K, T, u, d, N):
    """
    Args:
    - S_ini: Initial underlying price (S_0)
    - K: Call option's Strike price
    - T: Time to maturity (time horizone in periods)
    - u: upward price multiplier
    - d: downward price multiplier
    - N: Number of time steps.
    """
    C = np.zeros([N + 1, N + 1])  # Call payoff
    S = np.zeros([N + 1, N + 1])  # Underlying price
    for i in range(0, N + 1):
        C[N, i] = max(S_ini * (u ** (i)) * (d ** (N - i)) - K, 0) # Call Option payoff at terminal nodes.
        S[N, i] = S_ini * (u ** (i)) * (d ** (N - i)) # Price at terminal
    for j in range(N - 1, -1, -1):
        for i in range(0, j + 1):
            S[j, i] = S_ini * (u ** (i)) * (d ** (j - i))
    return S, C
```

```python
S_paths, Call = binomial_tree_call(100, 90, 3, 1.15, 0.85, 3)
print("Underlying Price Evolution:\n", S_paths)
print("\n Call Option Payoff:\n", Call)
```

#### **Risk-Neutral Probabilities**
#### One-step backward induction for valuation of Call Option:

- Risk-neutral probability is calculated at one-step tree: $$p = \frac{e^r_f \partial t - d}{u - d}$$
- At each node $[j,i]$ ($j$ steps with i steps up and $j-i$ steps down): the option price is computed backward: discounted from its expected price in the next steps:
$$ C_{j, i} = e^{-r_f \partial t} \left[ p C_{j+1, i+1} + (1-p) C_{j + 1, i} \right]$$

```python
def binomial_call_full(S_ini, K, T, r, u, d, N):
    """
    Args:
    - S_ini: Initial underlying price (S_0)
    - K: Call option's Strike price
    - T: Time to maturity (time horizone in periods)
    - r: risk-free rate
    - u: upward price multiplier
    - d: downward price multiplier
    - N: Number of time steps.
    """
    dt = T / N  # Define time step to calculate discounted rate
    p = (np.exp(r * dt) - d) / (u - d)  # Risk neutral probabilities (probs)
    C = np.zeros([N + 1, N + 1])  # Call prices/payoff
    S = np.zeros([N + 1, N + 1])  # Underlying price
    for i in range(0, N + 1):
        C[N, i] = max(S_ini * (u ** (i)) * (d ** (N - i)) - K, 0) #start from call payoff at maturity
        S[N, i] = S_ini * (u ** (i)) * (d ** (N - i))
    for j in range(N - 1, -1, -1):
        for i in range(0, j + 1):
            C[j, i] = np.exp(-r * dt) * (p * C[j + 1, i + 1] + (1 - p) * C[j + 1, i]) # Call price at each non-terminal node.
            S[j, i] = S_ini * (u ** (i)) * (d ** (j - i))
    return C[0, 0], C, S
```

```python
call_price, C_paths, S_paths = binomial_call_full(100, 90, 10, 0, 1.15, 0.85, 3)
#print("Underlying Price Evolution:\n", S_paths)
#print("\n Call Option Payoff:\n", C_paths)
print("\n Call Option Price at t=0: ", "{:.2f}".format(call_price))
```

```python
# Below plot code snippet is created by AI
N = S_paths.shape[0] - 1

fig, ax = plt.subplots(figsize=(9, 5))

# Connect each node to its down and up successors
for j in range(N):
    for i in range(j + 1):
        price = S_paths[j, i]

        ax.plot(
            [j, j + 1],
            [price, S_paths[j + 1, i]],
            color="steelblue"
        )
        ax.plot(
            [j, j + 1],
            [price, S_paths[j + 1, i + 1]],
            color="steelblue"
        )

# Draw and label valid nodes only
for j in range(N + 1):
    for i in range(j + 1):
        price = S_paths[j, i]
        call_p = C_paths[j,i]
        ax.scatter(j, price, color="navy", zorder=3)
        ax.annotate(
            f"{price:.2f} \n ({call_p:.2f})",
            (j, price),
            xytext=(5, 5),
            textcoords="offset points"
        )

ax.set(
    title="Underlying Price Binomial Tree",
    xlabel="Time step",
    ylabel="Underlying price"
)
ax.set_xticks(range(N + 1))
ax.margins(x=0.12, y=0.15)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## **5.2 DELTA HEDGING**


Consider shorting the following option:
- Call option
- Strike: $K = \$ 90$;

- Expiry: $T = 1$ year;
- Underlying initial price: $S_0 = 100\$$

The underlying price follows the evolution of Binomial Tree with 3 time steps:

```python
def binomial_call_delta(S_ini, K, T, r, u, d, N):
    """
    Args:
    - S_ini: Initial underlying price (S_0)
    - K: Call option's Strike price
    - T: Time to maturity (time horizone in periods)
    - r: risk-free rate
    - u: upward price multiplier
    - d: downward price multiplier
    - N: Number of time steps.
    """
    dt = T / N  # Define time step to calculate discounted rate
    p = (np.exp(r * dt) - d) / (u - d)  # Risk neutral probabilities (probs)
    C = np.zeros([N + 1, N + 1])  # Call prices/payoff
    S = np.zeros([N + 1, N + 1])  # Underlying price
    Delta = np.zeros([N, N])  # Delta values
    for i in range(0, N + 1):
        C[N, i] = max(S_ini * (u ** (i)) * (d ** (N - i)) - K, 0) #start from call payoff at maturity
        S[N, i] = S_ini * (u ** (i)) * (d ** (N - i))
    for j in range(N - 1, -1, -1):
        for i in range(0, j + 1):
            C[j, i] = np.exp(-r * dt) * (p * C[j + 1, i + 1] + (1 - p) * C[j + 1, i]) # Call price at each non-terminal node.
            S[j, i] = S_ini * (u ** (i)) * (d ** (j - i))
            Delta[j, i] = (C[j + 1, i + 1] - C[j + 1, i]) / (S[j + 1, i + 1] - S[j + 1, i]) #------------> Delta at each non-terminal node.
    return C[0, 0], C, S, Delta
```

```python
call_price, C_paths, S_paths, Delta_paths = binomial_call_delta(100, 90, 1, 0, 1.15, 0.85, 3)
#print("Underlying Price Evolution:\n", S_paths)
#print("\n Call Option Payoff:\n", C_paths)
print("\n Call Option Price at t=0: ", "{:.2f}".format(call_price))
#print("\n Delta at t=0: ", "{:.3f}".format(Delta_paths[0, 0]))
```

## **5.3. MATCHING VOLATILITY**

Consider for the underlying price evolutions:
- $T$ = 1 years
- Discretized in $N$ time steps, each step in the time interval $\Delta t = \frac{T}{N}$
- Annualized volatility of the underlying asset's return is $\sigma = 0.3 (= 30\%)$

**Recall**:
$$ u = e^{\sigma \sqrt{\Delta t}} \quad, \quad d = e^{-\sigma \sqrt{\Delta t}}$$

```python
def binomial_tree_vol(S_ini, T, N, sigma, r):
    """
    Build a recombining binomial price tree.

    Args:
        S_ini: Initial underlying price.
        T: Time horizon in years.
        N: Positive integer number of time steps.
        sigma: Annualized volatility (e.g., 0.20 for 20%).

    Returns:
        S: Array where S[j, i] is the price at step j
           after i up moves and j - i down moves.
           Entries outside the tree remain zero.
    """
    if not isinstance(N, (int, np.integer)) or N <= 0:
        raise ValueError("N must be a positive integer.")
    if T < 0 or sigma < 0:
        raise ValueError("T and sigma must be nonnegative.")

    dt = T / N
    u = np.exp(sigma * np.sqrt(dt))
    d = 1.0 / u
    p = (np.exp(r * dt) - d) / (u - d)
    S = np.zeros((N + 1, N + 1))
    for j in range(N + 1):
        i = np.arange(j + 1)
        S[j, :j + 1] = S_ini * u**i * d**(j - i)

    return S, p
```

```python
T = 1
sigma = 0.3
S0 = 100
N = 200
r = 0.03

S, p = binomial_tree_vol(S0, T, N, sigma, r)
```

```python
n_paths = 50
rng = np.random.default_rng(42)
time = np.linspace(0, T, N + 1)

plt.figure(figsize=(10, 6))

for _ in range(n_paths):
    up_moves = rng.binomial(1, p, size=N)
    up_count = np.concatenate(([0], np.cumsum(up_moves)))

    prices = S[np.arange(N + 1), up_count]
    plt.plot(time, prices, linewidth=1.2)

plt.xlabel("Time (years)")
plt.ylabel("Underlying price")
plt.title("Price evolution as $\Delta t \mapsto \infty$")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```
